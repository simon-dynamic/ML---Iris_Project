```python
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import LabelEncoder, StandardScaler
```

```python
url = "https://raw.githubusercontent.com/uiuc-cse/data-fa14/gh-pages/data/iris.csv"
df = pd.read_csv(url)
print(" Data set Head values ")
print(df.head())
```

```python
#Exploratory Data Analysis 

print("shape", df.shape)
print("info", df.info())
print("statistical summary", df.describe())
print(df["species"].value_counts())
```

```python
#Checking for null values
print(" ---- MISSING VALUES SUMMARY ---")
print(df.isnull().sum())
```

```python
# Added null values intentionally
df.iloc[2,0] = np.nan
df.iloc[5,1] = np.nan 
print(" Rechecking for null values")
missing_count = df.isnull().sum()
print(missing_count)
```

```python
if df.isnull().sum().sum()>0:
    print( " Found null values , replacing with mean/median")
    df["sepal_length"] = df["sepal_length"].fillna(df["sepal_length"].mean())
    df["sepal_width"] = df["sepal_width"].fillna(df["sepal_width"].mean())


print(" after imputation null count is:\n ", df.isnull())
```

```python
# Encoding categorical values ..
le = LabelEncoder()
df['species_encoded']=le.fit_transform(df['species'])
print('Label encoded data preview')
print(df[['species', 'species_encoded']].drop_duplicates())
```

```python
# Feature selection
#input features x and target output y
x = df[['sepal_length', 'sepal_width','petal_length','petal_width']]
y = df['species_encoded']
print("input features x shape : ", x.shape)
print("target output y shape : ", y.shape)
```

```python
# Feature normalization (min - max scaling)
x_normalized = (x - x.min()) / (x.max() - x.min())

print("\n=== 5. Normalized Features (First 5 Rows) ===")
print(x_normalized.head())
```
